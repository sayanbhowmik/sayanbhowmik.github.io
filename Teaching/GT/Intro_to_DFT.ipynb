# Introduction to Density Functional Theory

Following contains the pre-recorded lectures which were used for the Vertically Integrated Projects: Big Data and Quantum Mechanics class. The lectures on DFT were recorded in the year 2024 when I served as the instructor for the first time in this course. I have taught this course in different capacities including subteam lead from 2024 - 2027 in the Fall and Spring semesters. Dr. A. J. Medford has been running this course with his graduate students and postdocs serving as instructors at Georgia Tech since 2017. More can be found about this course at [VIP-BDQM](https://www.medford.chbe.gatech.edu/training-materials/docs/VIP_syllabus.html#).

Students ranged from Undergraduate research students to graduate students in various online MS programs run at GT.
