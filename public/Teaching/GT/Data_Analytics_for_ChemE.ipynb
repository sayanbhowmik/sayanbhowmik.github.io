# Data Analytics for Chemical Engineers (ChBE 4745/6745)

The primary objective of this course was to introduce data science and machine learning techniques for analyzing, interpreting, and visualizing datasets commonly encountered in chemical and biomolecular engineering. The course was aimed at undergraduate seniors and graduate students.

I was a TA for this course, taught by my PhD advisor Dr. A. J. Medford in Fall 2025. When I was a student in this course in Fall 2024, Dr. Medford piloted a new format of final examination - "Write your own problem (WYOP)" in response to improving Large Language Models and chatbots for coding tasks. I returned in the subsequent Fall as a Teaching Assistant and helped develop the pattern and rubric more rigorously for WYOP. All the exams - midterm and final - were replaced with WYOP-style exams.

Georgia Tech released [The Georgia Tech Academic AI Playbook: 2026 Edition](https://sites.gatech.edu/academic-ai-playbook/) - a cool resource pooled across the institute documenting how the faculty reacted to AI in teaching & research. You can read more about our contribution regarding the [WYOP approach](https://c14750.wpmucdn.com/5843/files/2026/09/gt-ai-playbook-2026-241-244-medford-yu-bhowmik-replacing-exams-with-write.pdf) in the 2026 edition of the playbook.

My other responsibilities included conducting office hours and advising students on the course-long project. I had won the "Best Online TA Award" for the department from the Center for Teaching and Learning, GT and was recognized in the institute-wide ceremony. As always, my real reward was the student feedback at the end of the course where a student remarked, "Sayan was an absolute beast; he was super helpful and responsive when I had questions for him" and would "buy him a beer if I saw him out on a weekend," while another student "truly valued the Python session at the beginning of the semester".
